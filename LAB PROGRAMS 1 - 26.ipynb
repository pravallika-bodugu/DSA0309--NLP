**Program 2: Finite State Automaton (FSA)**

Aim: To implement a basic finite state automaton in Python that recognizes strings ending with ab

```python
# Finite State Automaton to recognize strings ending with 'ab'

def check_string(s):
    state = 0

    for char in s:
        if state == 0:
            if char == 'a':
                state = 1
            else:
                state = 0

        elif state == 1:
            if char == 'a':
                state = 1
            elif char == 'b':
                state = 2
            else:
                state = 0

        elif state == 2:
            if char == 'a':
                state = 1
            else:
                state = 0

    if state == 2:
        return "Accepted"
    else:
        return "Rejected"


# Test strings
strings = ["ab", "aab", "abab", "abc", "ba"]

for s in strings:
    print(s, ":", check_string(s))
```

**Program 4:Finite-State Machine for Morphological Parsing**

Aim:To implement a finite-state machine in Python to generate plural forms of English nouns.

This program uses simple rules to convert singular nouns into plural nouns.

**Examples:**
- cat → cats
- box → boxes
- baby → babies

```python
# Finite-State Machine for Plural Formation

def plural(noun):
    state = "START"

    # Check noun ending
    if noun.endswith(("s", "sh", "ch", "x", "z")):
        state = "ADD_ES"

    elif noun.endswith("y") and noun[-2].lower() not in "aeiou":
        state = "CHANGE_Y_TO_IES"

    else:
        state = "ADD_S"

    # Apply the corresponding rule
    if state == "ADD_ES":
        return noun + "es"

    elif state == "CHANGE_Y_TO_IES":
        return noun[:-1] + "ies"

    elif state == "ADD_S":
        return noun + "s"


# Test nouns
nouns = ["cat", "box", "bus", "baby", "book"]

print("Plural Forms of Nouns:\n")

for noun in nouns:
    print(noun, "->", plural(noun))
```

**Program 6: N-gram Model for Text Generation**

**Aim:** To implement a basic N-gram model for text generation using a Bigram Model in Python.

**N-gram:** An N-gram is a sequence of N consecutive words or characters.
- Unigram: One word
- Bigram: Two consecutive words
- Trigram: Three consecutive words
A Bigram Model predicts the next word based on the previous word.

```python
import re
from collections import defaultdict, Counter

# Sample training text
text = "I like NLP. I like Python. NLP is interesting."

# Convert text to lowercase and tokenize
words = re.findall(r'\b\w+\b', text.lower())

# Create Bigram Model
bigrams = defaultdict(Counter)

for i in range(len(words) - 1):
    bigrams[words[i]][words[i + 1]] += 1

# Generate text
current_word = "i"
generated_text = [current_word]

for i in range(5):
    if current_word in bigrams:
        next_word = bigrams[current_word].most_common(1)[0][0]
        generated_text.append(next_word)
        current_word = next_word
    else:
        break

print("Original Text:", text)
print("Generated Text:", " ".join(generated_text))
```

**Program 8: Stochastic POS Tagging using a Probabilistic Model**

Aim: To implement a simple stochastic Part-of-Speech (POS) tagging algorithm using a probabilistic model to assign POS tags to words.

Stochastic POS tagging uses probabilities to determine the most likely grammatical tag for a word.

**For example:**
- dog → Noun (NN)
- run → Verb (VB)
- beautiful → Adjective (JJ)
In this program, we use a simple Hidden Markov Model (HMM)-style approach with word emission probabilities and tag transition probabilities.

```python
import math
from collections import defaultdict, Counter

# Training data: words with their POS tags
training_data = [
    [("the", "DET"), ("dog", "NOUN"), ("barks", "VERB")],
    [("the", "DET"), ("cat", "NOUN"), ("runs", "VERB")],
    [("a", "DET"), ("dog", "NOUN"), ("runs", "VERB")],
    [("a", "DET"), ("cat", "NOUN"), ("barks", "VERB")]
]

# Calculate emission and transition probabilities
emission = defaultdict(Counter)
transition = defaultdict(Counter)
tag_counts = Counter()

for sentence in training_data:
    previous = "START"

    for word, tag in sentence:
        emission[tag][word] += 1
        transition[previous][tag] += 1
        tag_counts[tag] += 1
        previous = tag

tags = list(tag_counts.keys())

# Stochastic POS tagging
def pos_tag(sentence):
    words = sentence.lower().split()
    dp = [{}]
    backpointer = [{}]

    for tag in tags:
        start_total = sum(transition["START"].values())
        trans = transition["START"][tag] / start_total if start_total > 0 else 0
        emit = emission[tag][words[0]] / tag_counts[tag]
        dp[0][tag] = math.log(trans) + math.log(emit) if emit and trans else -float("inf")

    for i in range(1, len(words)):
        dp.append({})
        backpointer.append({})

        for tag in tags:
            emit = emission[tag][words[i]] / tag_counts[tag]
            best_score = -float("inf")
            best_previous = None

            for prev in tags:
                prev_total = sum(transition[prev].values())
                trans = transition[prev][tag] / prev_total if prev_total > 0 else 0
                score = dp[i-1][prev] + math.log(trans) if trans else -float("inf")

                if score > best_score:
                    best_score = score
                    best_previous = prev

            dp[i][tag] = best_score + math.log(emit) if emit else -float("inf")
            backpointer[i][tag] = best_previous

    # Find the best tag sequence
    last_tag = max(dp[-1], key=dp[-1].get)
    result = [last_tag]

    for i in range(len(words)-1, 0, -1):
        last_tag = backpointer[i][last_tag]
        result.append(last_tag)

    return list(zip(words, reversed(result)))

# Test sentence
sentence = "the dog barks"

print("Stochastic POS Tagging:")
for word, tag in pos_tag(sentence):
    print(word, "->", tag)
```

**Program 9: Rule-Based POS Tagging using Regular Expressions**

Aim: To implement a rule-based Part-of-Speech (POS) tagging system using regular expressions in Python.

Rule-based POS tagging assigns grammatical tags to words based on predefined rules and patterns.

**Examples:**
- Words ending in ing → Verb
- Words ending in ly → Adverb
- Words ending in ous → Adjective
- Words ending in s → Noun (in this simplified example)

```python

import nltk
from nltk.tokenize import word_tokenize
from nltk.tag import RegexpTagger

# Download tokenizer
nltk.download('punkt_tab')

# Define regular expression rules
patterns = [
    (r'^(The|the|A|a|An|an)$', 'DT'),
    (r'^(is|am|are|was|were|be)$', 'VB'),
    (r'.*ing$', 'VBG'),
    (r'.*ed$', 'VBD'),
    (r'.*ly$', 'RB'),
    (r'.*(ous|ful|able|ive)$', 'JJ'),
    (r'^(in|on|at|over|under|with)$', 'IN'),
    (r'^(dog|cat|fox|book|student)$', 'NN'),
    (r'.*', 'NN')
]

# Create rule-based tagger
tagger = RegexpTagger(patterns)

# Input sentence
text = "The quick fox jumps over the lazy dog"

# Tokenize the sentence
words = word_tokenize(text)

# Perform POS tagging
tags = tagger.tag(words)

# Display results
print("Rule-Based POS Tagging:\n")

for word, tag in tags:
    print(word, "->", tag)
```

**Program 10: Transformation-Based Tagging**

Aim: To implement transformation-based tagging using a set of transformation rules and apply a simple rule to tag words in Python.

Transformation-based tagging is a method that initially assigns tags to words and then improves them by applying correction rules.

**For example:**
- Initial tag: playing → Noun
- Transformation rule: Words ending with ing are verbs.
- Corrected tag: playing → Verb

```python

# Transformation-Based Tagging

# Input sentence
sentence = "The boy is playing and running"

# Initial tagging
words = sentence.split()

tags = []

for word in words:
    tags.append((word, "NN"))

print("Initial Tagging:")
print(tags)

# Transformation rule:
# Words ending with 'ing' are tagged as verbs (VBG)

transformed_tags = []

for word, tag in tags:
    if word.lower().endswith("ing"):
        tag = "VBG"

    transformed_tags.append((word, tag))

print("\nAfter Transformation:")
for word, tag in transformed_tags:
    print(word, "->", tag)
```

**Program 11: Top-Down Parser for Context-Free Grammar (CFG)**

Aim: To implement a simple top-down parser for a Context-Free Grammar (CFG) using Python.

A Context-Free Grammar (CFG) defines the structure of sentences using grammar rules.

A top-down parser starts with the start symbol and tries to derive the input sentence using grammar rules.

```python

import nltk
from nltk import CFG
from nltk.parse import RecursiveDescentParser

# Define Context-Free Grammar
grammar = CFG.fromstring("""
S -> NP VP
NP -> Det N
VP -> V NP
Det -> 'the' | 'a'
N -> 'dog' | 'cat'
V -> 'chased' | 'saw'
""")

# Create top-down parser
parser = RecursiveDescentParser(grammar)

# Input sentence
sentence = "the dog chased the cat".split()

# Parse the sentence
print("Parse Tree:\n")

for tree in parser.parse(sentence):
    tree.pretty_print()
```

**Program 14: Agreement Checking using Context-Free Grammar**

Aim: To create a Python program to check for agreement in sentences based on Context-Free Grammar (CFG) rules.

Agreement means that words in a sentence must match grammatically, especially in number (singular or plural).

**Examples:**
- The boy runs. — Correct (singular subject + singular verb)
- The boys run. — Correct (plural subject + plural verb)
- The boy run. — Incorrect (singular subject + plural verb)

```python

import nltk
from nltk import CFG
from nltk.parse import ChartParser

# Define grammar with subject-verb agreement
grammar = CFG.fromstring("""
S -> NP_S VP_S | NP_P VP_P

NP_S -> Det N_S
NP_P -> Det N_P

VP_S -> V_S
VP_P -> V_P

Det -> 'the'

N_S -> 'boy' | 'girl'
N_P -> 'boys' | 'girls'

V_S -> 'runs' | 'plays'
V_P -> 'run' | 'play'
""")

# Create parser
parser = ChartParser(grammar)

# Test sentences
sentences = [
    "the boy runs",
    "the boys run",
    "the boy run",
    "the girls plays"
]

for sentence in sentences:
    words = sentence.split()

    trees = list(parser.parse(words))

    if trees:
        print(sentence, "-> Correct Agreement")
    else:
        print(sentence, "-> Incorrect Agreement")
```

**Program 15: Probabilistic Context-Free Grammar (PCFG) Parsing**

Aim: To implement probabilistic Context-Free Grammar (PCFG) parsing for a sentence using Python.

A PCFG is a Context-Free Grammar in which each grammar rule is assigned a probability. The parser uses these probabilities to identify the most likely grammatical structure of a sentence.

```python

import nltk
from nltk import PCFG
from nltk.parse import ViterbiParser

# Define Probabilistic Context-Free Grammar
grammar = PCFG.fromstring("""
S -> NP VP [1.0]
NP -> Det N [0.6] | NP PP [0.4]
VP -> V NP [0.7] | VP PP [0.3]
PP -> P NP [1.0]
Det -> 'the' [1.0]
N -> 'dog' [0.5] | 'cat' [0.5]
V -> 'chased' [1.0]
P -> 'with' [1.0]
""")

# Create probabilistic parser
parser = ViterbiParser(grammar)

# Input sentence
sentence = "the dog chased the cat".split()

# Parse the sentence
print("PCFG Parse Tree:\n")

for tree in parser.parse(sentence):
    tree.pretty_print()
    print("Probability:", tree.prob())
```

**Program 18: First-Order Predicate Calculus (FOPC) Parser**

Aim: To implement a simple First-Order Predicate Calculus (FOPC) parser for basic logical expressions using Python.

FOPC is a formal language used to represent facts, objects, relationships, and logical statements.

**Examples of logical expressions:**
- P(x) – Predicate P applied to x
- P(x) AND Q(x) – Conjunction
- P(x) OR Q(x) – Disjunction
- NOT P(x) – Negation
This program recognizes simple predicate expressions using regular expressions.

```python

import re

# Tokenize logical expressions
def tokenize(expression):
    pattern = r'AND|OR|NOT|[A-Za-z_]\w*\([^()]*\)|[()]'
    return re.findall(pattern, expression)

# Recursive descent parser
class FOPCParser:
    def __init__(self, tokens):
        self.tokens = tokens
        self.pos = 0

    def match(self, expected):
        if self.pos < len(self.tokens) and self.tokens[self.pos] == expected:
            self.pos += 1
            return True
        return False

    def parse_expression(self):
        left = self.parse_term()

        while self.match("OR"):
            right = self.parse_term()
            left = ("OR", left, right)

        return left

    def parse_term(self):
        left = self.parse_factor()

        while self.match("AND"):
            right = self.parse_factor()
            left = ("AND", left, right)

        return left

    def parse_factor(self):
        if self.match("NOT"):
            return ("NOT", self.parse_factor())

        if self.match("("):
            result = self.parse_expression()
            if not self.match(")"):
                raise ValueError("Missing closing parenthesis")
            return result

        if self.pos < len(self.tokens):
            token = self.tokens[self.pos]

            if re.fullmatch(r'[A-Za-z_]\w*\([^()]*\)', token):
                self.pos += 1
                return ("PREDICATE", token)

        raise ValueError("Invalid logical expression")

    def parse(self):
        result = self.parse_expression()

        if self.pos != len(self.tokens):
            raise ValueError("Unexpected token")

        return result

# Input logical expression
expression = "Human(x) AND NOT Animal(x)"

try:
    tokens = tokenize(expression)
    parser = FOPCParser(tokens)
    result = parser.parse()

    print("Logical Expression:", expression)
    print("Tokens:", tokens)
    print("Parsed Expression:", result)
    print("Expression is valid")

except ValueError as e:
    print("Invalid Expression:", e)
```

**Program 20: Information Retrieval using TF-IDF**

Aim: To implement a basic Information Retrieval System using TF-IDF (Term Frequency–Inverse Document Frequency) for document ranking in Python.

Information Retrieval (IR) is the process of finding relevant documents based on a user's query.

**TF-IDF assigns importance to words based on:**
- TF (Term Frequency): How frequently a word occurs in a document.
- IDF (Inverse Document Frequency): How rare or important a word is across all documents.

```python
!pip install scikit-learn
```

```python

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Sample documents
documents = [
    "Python is a programming language",
    "Natural Language Processing uses Python",
    "Machine learning is a branch of artificial intelligence",
    "Python is useful for machine learning"
]

# User query
query = "Python machine learning"

# Create TF-IDF vectorizer
vectorizer = TfidfVectorizer()

# Convert documents into TF-IDF vectors
tfidf_matrix = vectorizer.fit_transform(documents)

# Convert query into TF-IDF vector
query_vector = vectorizer.transform([query])

# Calculate similarity scores
scores = cosine_similarity(query_vector, tfidf_matrix)[0]

# Rank documents by similarity
ranked_docs = sorted(
    enumerate(scores),
    key=lambda x: x[1],
    reverse=True
)

# Display results
print("Information Retrieval Results:\n")

for index, score in ranked_docs:
    print("Document:", documents[index])
    print("Similarity Score:", round(score, 3))
    print()
```

**Program 21: Syntax-Driven Semantic Analysis**

Aim: To perform syntax-driven semantic analysis by extracting noun phrases from a sentence and identifying their meanings using Python and NLTK.
What is Syntax-Driven Semantic Analysis?
- Syntax analysis: Identifies grammatical structures in a sentence.
- Semantic analysis: Determines the meaning of words or phrases.
- This program identifies noun phrases and retrieves their meanings from WordNet.

```python
import nltk

nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('wordnet')
```

```python
import nltk
from nltk import word_tokenize, pos_tag
from nltk.chunk import RegexpParser
from nltk.corpus import wordnet

# Input sentence
sentence = "The dog chased the cat."

# Tokenize the sentence
tokens = word_tokenize(sentence)

# Perform POS tagging
tagged_words = pos_tag(tokens)

print("POS Tagged Words:")
print(tagged_words)

# Define grammar for noun phrases
grammar = r"""
    NP: {<DT>?<JJ.*>*<NN.*>+}
"""

# Create parser
parser = RegexpParser(grammar)

# Parse the sentence
tree = parser.parse(tagged_words)

print("\nParse Tree:")
tree.pretty_print()

# Extract noun phrases and meanings
print("\nNoun Phrases and Meanings:")

for subtree in tree.subtrees(
    filter=lambda t: t.label() == "NP"
):
    phrase = " ".join(word for word, tag in subtree.leaves())

    # Find the noun in the phrase
    noun = next(
        (word for word, tag in reversed(subtree.leaves())
         if tag.startswith("NN")),
        None
    )

    print("\nNoun Phrase:", phrase)

    if noun:
        synsets = wordnet.synsets(noun, pos=wordnet.NOUN)

        if synsets:
            print("Meaning:", synsets[0].definition())
        else:
            print("Meaning not found")
```

**Program 22: Reference Resolution within Text**

Aim: To identify the noun or person referred to by pronouns such as he, she, it, and they using Python and NLTK.
What is Reference Resolution?
Reference resolution is an NLP technique used to identify the entity to which a pronoun refers.

**Example:**

Sentence: “Ravi bought a laptop. He uses it daily.”
- “He” refers to Ravi.
- “It” refers to the laptop.

```python
import nltk

nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
import nltk
from nltk import word_tokenize, pos_tag

# Input text
text = "Ravi bought a laptop. He uses it daily."

# Tokenize text
tokens = word_tokenize(text)

# Perform POS tagging
tagged_words = pos_tag(tokens)

print("POS Tagged Words:")
print(tagged_words)

# Extract nouns
nouns = [
    word for word, tag in tagged_words
    if tag.startswith("NN")
]

print("\nNouns:", nouns)

# Reference resolution
pronouns = ["he", "she", "it", "they"]

print("\nReference Resolution:")

for i, (word, tag) in enumerate(tagged_words):

    if word.lower() in pronouns:

        # Find the nearest preceding noun
        preceding_nouns = [
            w for w, t in tagged_words[:i]
            if t.startswith("NN")
        ]

        if preceding_nouns:
            reference = preceding_nouns[-1]
            print(word, "refers to", reference)
        else:
            print(word, "reference not found")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Input text
text = """
It was raining heavily.
I took an umbrella to protect myself.
The weather was cold and cloudy.
I stayed inside my house.
"""

# Split text into sentences
sentences = [
    sentence.strip()
    for sentence in text.strip().split("\n")
    if sentence.strip()
]

print("Sentences:")
for sentence in sentences:
    print(sentence)

# Convert sentences into TF-IDF vectors
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(sentences)

# Calculate similarity between consecutive sentences
similarities = []

print("\nSentence Similarity:")

for i in range(len(sentences) - 1):

    similarity = cosine_similarity(
        tfidf_matrix[i],
        tfidf_matrix[i + 1]
    )[0][0]

    similarities.append(similarity)

    print(
        f"Sentence {i + 1} and Sentence {i + 2}: "
        f"{similarity:.2f}"
    )

# Calculate coherence score
if similarities:
    coherence_score = sum(similarities) / len(similarities)

    print("\nCoherence Score:", round(coherence_score, 2))
```

**Program 24: Recognize Dialog Acts in Conversation**

Aim: To recognize dialog acts in a conversation using Python and Natural Language Processing techniques.
What is Dialog Act Recognition?
Dialog act recognition identifies the purpose or intention of a sentence in a conversation.

**Examples:**
- “Hello!” → Greeting
- “What is your name?” → Question
- “Please help me.” → Request
- “Thank you.” → Thanks
- “Goodbye!” → Farewell

```python
# Dialog Act Recognition

def recognize_dialog_act(sentence):

    sentence = sentence.lower().strip()

    if any(word in sentence for word in ["hello", "hi", "hey"]):
        return "Greeting"

    elif sentence.startswith(("what", "where", "when", "why", "how", "who")):
        return "Question"

    elif "please" in sentence or "help" in sentence:
        return "Request"

    elif "thank" in sentence:
        return "Thanks"

    elif any(word in sentence for word in ["bye", "goodbye", "see you"]):
        return "Farewell"

    else:
        return "Statement"


# Input conversation
conversation = [
    "Hello!",
    "What is your name?",
    "Please help me.",
    "Thank you.",
    "Goodbye!",
    "Python is a programming language."
]

# Recognize dialog acts
print("Dialog Act Recognition:\n")

for sentence in conversation:
    act = recognize_dialog_act(sentence)

    print("Sentence:", sentence)
    print("Dialog Act:", act)
    print()
```

**Program 26: Machine Translation Using Hugging Face Transformers**

Aim: To translate English text into French using Python and the Hugging Face Transformers library.
What is Machine Translation?
Machine translation is the process of automatically translating text from one language to another using a computer.

**Example:**
- English: “How are you?”
- French: “Comment allez-vous ?”
Hugging Face provides pretrained models that can perform language translation.

```python
!pip install -q transformers sentencepiece
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# Load pretrained translation model
model_name = "Helsinki-NLP/opus-mt-en-fr"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

# English input
english_text = "How are you?"

# Tokenize input
inputs = tokenizer(english_text, return_tensors="pt")

# Generate translation
outputs = model.generate(**inputs)

# Convert tokens into French text
french_text = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print("English:", english_text)
print("French:", french_text)
```
